# Logboek Maartje casus B: image processing

## Inleiding

Dit is een persoonlijk logboek voor de 2e casus van het vak Modelleren van Kanker. In deze casus gaan we bezig met image processing, wat in drie opdrachten uitgewerkt wordt. Om te beginnen start de eerste opdracht met een artificieel neuraal netwerk voor beeldherkenning. De tweede opdracht zal gaan over het bepalen van tumoren in MRI-scans. De derde opdracht gaat over een soort van YOLO.
